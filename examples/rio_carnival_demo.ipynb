# carnaval_viz demo

This notebook demonstrates `carnaval_viz` using the Rio Carnival Blocos dataset (see `prepare_dataset.py` for how the raw file is cleaned).

```python
import pandas as pd
import carnaval_viz as viz
from prepare_dataset import CLEAN_PATH, prepare_dataframe

if not CLEAN_PATH.exists():
    prepare_dataframe().to_csv(CLEAN_PATH, index=False)
df = pd.read_csv(CLEAN_PATH)
df.head()
```

## Bubble chart

Each bubble is one bloco: founding year on the x-axis, estimated audience on the y-axis (log-scaled) and as bubble size (square-root scaled), region as color.

```python
bubble_fig = viz.bubble_chart(
    df,
    x="year_founded",
    y="estimated_audience",
    size="estimated_audience",
    color="region",
    annotate_top=5,
)
bubble_fig
```

## Carnival calendar

Every event placed around a calendar wheel scoped to the Carnival season: angle is the event date, radius is the event's start time, size is audience, color is region.

```python
calendar_fig = viz.carnival_calendar(
    df,
    date="event_date",
    time="gathering_time",
    size="estimated_audience",
    color="region",
)
calendar_fig
```

## Saving figures

Both functions return a Matplotlib `Figure`, so you can save them like any other figure.

```python
bubble_fig.savefig("bubble_chart.png", dpi=300, bbox_inches="tight")
calendar_fig.savefig("carnival_calendar.png", dpi=300, bbox_inches="tight")
```

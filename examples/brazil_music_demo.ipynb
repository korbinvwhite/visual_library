# carnaval_viz demo

This notebook demonstrates `carnaval_viz` using a synthetic, Brazilian-music-style example dataset (see `generate_dataset.py` for how it's generated).

```python
import carnaval_viz as viz
from generate_dataset import generate_sample_dataframe

df = generate_sample_dataframe()
df.head()
```

## Histogram

Plot the distribution of one numeric column, with mean and median reference lines.

```python
hist_fig = viz.histogram(df, "danceability")
hist_fig
```

## Correlation heatmap

Plot pairwise correlations across all numeric audio features.

```python
corr_fig = viz.correlation(df)
corr_fig
```

## Saving figures

Both functions return a Matplotlib `Figure`, so you can save them like any other figure.

```python
hist_fig.savefig("danceability_histogram.png", dpi=300, bbox_inches="tight")
corr_fig.savefig("music_correlation.png", dpi=300, bbox_inches="tight")
```
